```python
import requests

URL = "https://rest.isric.org/soilgrids/v2.0/properties/query"

lat, lon = 27.71, 85.32

params = {
    "lat": lat,
    "lon": lon,
    "property": "phh2o",
    "depth": "0-5cm",
    "value": "mean",
}

req = requests.get(URL, params=params, timeout=30)
req.raise_for_status() 

data = req.json()
print(data)
```
