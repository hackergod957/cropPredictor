```python
import pandas, requests, sys
print(sys.executable)
print(pandas.__version__)
```

```python
BASE = "https://power.larc.nasa.gov/api/temporal/climatology/point"

lat, lon = 27.71, 85.32

params = {
    "parameters" :",".join(["T2M", "PRECTOTCORR", "ALLSKY_SFC_SW_DWN", "RH2M"]),
    "community" : "AG",
    "latitude" : lat,
    "longitude" : lon,
    "format" : "JSON"
}

req = requests.get(BASE,params = params , timeout = 30)
```

```python
print(req.status_code,req.url)
```

```python
data = req.json()
print(list(data.keys()))
```
